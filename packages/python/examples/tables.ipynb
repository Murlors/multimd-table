# Edit and display MultiMarkdown tables

Install `multimd-table[markdown-it]` in this notebook's Python environment. To use a source checkout, run `%pip install "..[markdown-it]"` from this directory instead.

The Python core parses and writes tables. The optional markdown-it-py adapter renders the complete document, including Markdown inside cells. HTML appears as code output; this does not change how Jupyter renders Markdown cells.

```python
from multimd_table import parse_table, stringify_table

source = '''[Quarterly **revenue**]
Region|Revenue
---|---:
North|10
^^|20
Total||'''
table = parse_table(source, rowspan=True)
assert table is not None
table['sections'][1]['rows'][0]['cells'][1]['value'] = '12'
written = stringify_table(table)
reparsed = parse_table(written, rowspan=True)
assert reparsed is not None
assert reparsed['sections'][1]['rows'][0]['cells'][1]['value'] == '12'
print(written)
```

```python
from IPython.display import HTML, display
from markdown_it import MarkdownIt
from multimd_table.markdown_it import multimd_table_plugin

md = MarkdownIt('commonmark').enable('table').use(multimd_table_plugin, rowspan=True)
document = '# Report\n\n' + written + '\n\nUpdated from Python.'
html = md.render(document)
assert 'rowspan="2"' in html and 'colspan="2"' in html
assert '<strong>revenue</strong>' in html
display(HTML(html))
```
